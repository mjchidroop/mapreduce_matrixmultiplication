#### Experiment 01
#### Matrix multiplication using MapReduce

```python
from collections import defaultdict

matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

vector = [1, 2, 3]
```

```python
# Mapper:
def mapper(matrix, vector):
  mapped = []
  for i in range(len(matrix)):
    for j in range(len(matrix[0])):
      product = matrix[i][j] * vector[j]
      mapped.append((i, product))
  return mapped

# Reducer:
def reducer(mapped):
  result = defaultdict(int)
  for key, value in mapped:
    result[key] += value
  return result
```

```python
mapped = mapper(matrix, vector)
reduced = reducer(mapped)

print("="*20)
print("Mapped Output:")
print("-"*20)
for item in mapped:
  print(item)
print("="*20)
print("="*20)
print("Final Result:")
print("-"*20)
print("(Matrix x Vector):")
for row in sorted(reduced):
  print(f"Row {row} : {reduced[row]}")
print("="*20)
```

```python
# intermediates:

import pandas as pd
grouped = defaultdict(list)
for k, v in mapped:
  grouped[k].append(v)

df = pd.DataFrame({
    "Key (row)":   list(grouped.keys()),
    "Mapped vals": [grouped[k] for k in grouped],
    "Sum (reduce)": [sum(grouped[k]) for k in grouped]
})

df
```
